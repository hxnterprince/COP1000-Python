```python
try:
    age = int(input("Enter your age: "))
    print(f"You are {age} years old.")

except ValueError:
    print("Invalid input. Please enter a valid integer for age.")
```

```python
try:
    num1 = int(input("Enter the first number: "))
    num2 = int(input("Enter the second number: "))
    result = num1 / num2
    print(f"Result: {result}")
except ValueError:
    print("Error: Input must be a valid number.")
except ZeroDivisionError:
    print("You cannot divide by zero.")
```

```python
try:
    grade = int(input("Enter your grade (0-100): "))
    if 0 <= grade <= 100:
        print("Grade is valid.")
    else:
        print("Grade must be between 0 and 100.")
except ValueError as e:
    print("Error: Nonnumeric input entered.")
```

```python
while True:
    try:
        age = int(input("Enter your age: "))
        if age > 0:
            print(f"Age accepted: {age}")
            break
        else:
            print("Age must be greater than 0. Please try again.")
    except ValueError:
        print("Invalid input. Please enter a valid integer for age.")
```

```python
try:
    score1 = float(input("Enter the first score: "))
    score2 = float(input("Enter the second score: "))
    score3 = float(input("Enter the third score: "))

    average = (score1 + score2 + score3) / 3
    print(f"The average score is: {average:.2f}")
except ValueError:
    print("Invalid input. Please enter valid numeric scores.")
```
