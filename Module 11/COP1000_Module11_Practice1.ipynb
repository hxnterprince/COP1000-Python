```python
import math

print(f"The value of pi: {math.pi}")
print(f"The square root of 81: {math.sqrt(81)}")
print(f"5 raised to the power of 2: {math.pow(5, 2)}")
```

```python
import math

radius = float(input("Enter the radius: "))

area = math.pi * (radius ** 2)

print(f"Area: {area:.2f}")
```

```python
import random

print("Generating three random whole numbers from 1 through 10:")

for i in range(3):
    num = random.randint(1, 10)
    print(f"Random number {i+1}: {num}")
```

```python
import random

outcomes = ["Heads", "Tails"]

coin_flip = random.choice(outcomes)

print(f"The coin landed on: {coin_flip}")
```

```python
import random

for i in range(1, 6):
    roll = random.randint(1, 6)
    print(f"Roll {i}: {roll}")
```
