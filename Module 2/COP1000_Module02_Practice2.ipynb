```python
first_name = input("Enter your first name ")

print(f"Hello, {first_name}! Welcome to python.")

city = input("Enter your city ")
favfood = input(" Enter your favorite food ")
```

```python
print(f"{first_name} lives in {city} and their favorite food is {favfood}")

age = input("Enter your age ")
current_age = int(age)
older_age = current_age + 1
print(f"You are currently {current_age} years old")
print(f"Next year you will be turning {older_age} years old")

W1 = input("Enter your favorite number ")
W2 = input("Enter another favorite number ")
W1int = int(W1)
W2int = int(W2)
sum = W1int + W2int
print(f"The sum of them would equal to {sum}")

item_name = input("Enter an item name ")
item_price = input("The price you beleieve it is ")
number_of_items = input("Enter the amount you want of these items ")
float_price = float(item_price)
int_quantity = int(number_of_items)
print("Purchase Summary")
print(f"Item: {item_name}")
print(f"Price: ${float_price}")
print(f"Quantity: {int_quantity}")
total_cost = float_price * int_quantity
print(f"Total Cost: ${total_cost}")

hours_worked = float(input("Enter the amount of hours you have worked this week "))
hourly_pay = float(input("Enter how much you get paid by the hour "))
gross_pay = hours_worked * hourly_pay
print(f"Gross Pay: ${gross_pay}")

```
