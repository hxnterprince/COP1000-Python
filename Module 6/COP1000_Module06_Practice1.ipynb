```python
shopping = ["milk", "bread", "eggs", "apples", "coffee"]

shopping.remove("eggs")

shopping.pop()

print("Updated shopping list:", shopping)

print("Number of remaining items:", len(shopping))
```

```python
names = ["Doug", "James", "Donny", "Sarah", "Todd"]

print("OG Names:", names)

names.sort()

print("Alphabetically listed names:", names)

names.reverse()

print("Reversed aplhabetically listed names:", names)
```

```python
scores = [88, 94, 76, 91, 85]

highest = max(scores)
lowest = min(scores)
total = sum(scores)
count = len(scores)

print("Highest score:", highest)
print("Lowest score:", lowest)
print("Total sum:", total)
print("Number of scores:", count)

average = total / count

print(f"Average score: {average:.2f}")


```
