```python
temperatures = [45, 72, 88, 61, 93, 55, 79]

count = 0

print("Temp 70 or highier:")
for temp in temperatures:
    if temp >= 70:
        print(temp)
        count += 1
print(f"Total: {count}")

print("Number of temps that met the requirement:", count)
```

```python
scores = []

for i in range (1,6):
    score = float(input(f"Enter your test score {i}:  "))
    scores.append(score)

print("Scores list:", scores)

total = sum(scores)
print("Total sum:", total)

average = total / len(scores)
print("Average:", average)

print("Highest score:", max(scores))
print("Lowest score:", min(scores))
```

```python
weekend = ("Saturday", "Sunday")

print("Complete tuple:", weekend)
print("First item:", weekend[0])
print("Second item:", weekend[-1])
print("Number of items:", len(weekend))

print("-------------Days on the weekend--------------")
for day in weekend:
    print(day)
```
