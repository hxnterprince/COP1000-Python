```python
colors = {"red", "blue", "green", "yellow"}

print("Colors Set:")
print(colors)

print("Number of items in colors set:")
print(len(colors))
```

```python
courses = {"COP 1000", "ENC 1101", "MAT 1033"}

courses.add("SLS 1101")

courses.remove("MAT 1033")

print("Final Courses Set:")
print(courses)
```

```python
names = ["Alex", "Jordan", "Alex", "Taylor", "Jordan", "Morgan"]

print("OG List:")
print(names)

names_set = set(names)

print("Converted")
print(names_set)

print(f"Number of values in the og list: {len(names)}")
print(f"Number of unique values in the set: {len(names_set)}")
```

```python
languages = {"Python", "Java", "C++", "JavaScript"}

user_language = input("Enter a language: ")

if user_language in languages:
    print(f"{user_language} is a programming language.")
else:
    print(f"{user_language} is not a programming language.")
```

```python
student_a = {"Python", "Gaming", "Music", "Travel"}
student_b = {"Python", "Music", "Photography", "Sports"}

common_interests = student_a.intersection(student_b)

print("Common Interests:")
for interest in common_interests:
    print(interest)
```

```python
unique_interests = student_a.union(student_b)

print("All unique interests")
print(unique_interests)
```

```python
python_students = {"Alex", "Jordan", "Taylor", "Morgan"}
web_students = {"Jordan", "Morgan", "Casey", "Riley"}

both_courses = python_students.intersection(web_students)
either_course = python_students.union(web_students)
python_only = python_students.difference(web_students)

print("Both courses:", both_courses)
print("Either course:", either_course)
print("Python only:", python_only)
```
