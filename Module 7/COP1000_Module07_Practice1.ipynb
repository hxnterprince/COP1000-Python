```python
student = {
    "name": "Jordan",
    "major": "Cybersecurity",
    "credits": 30
}

print("Entire Dictionary")
print(student)

print("Individual Values")
print(student["name"])
print(student["major"])
print(student["credits"])
```

```python
student["gpa"] = 3.8

print("Updated Dictionary")
print(student)
print()
print("GPA Value:")
print(student["gpa"])
```

```python
course = {
    "course": "COP 1000",
    "room": "A101",
    "students": 24
}

print("OG Course Dictionary")
print(course)

course["room"] = "B205"

course["students"] = 26

print("Updated Course Dictionary")
print(course)
```

```python
profile = {
    "name": "Taylor",
    "city": "Jacksonville",
    "age": 21,
    "favorite_color": "blue"
}

del profile["favorite_color"]

print("Updated Profile")
print(profile)

print("Number of key-values pairs:")
print(len(profile))
```

```python
inventory = {
    "notebooks": 12,
    "pens": 30,
    "folders": 8
}

if "pens" in inventory:
    print("We have pens in inventory")
else:
    print("We do not have pens in inventory")

if "markers" in inventory:
    print("We have markers in inventory")
else:
    print("We do not have markers in inventory")

```
