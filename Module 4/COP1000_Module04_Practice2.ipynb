```python
password = "python123"
guess = input("Enter the password: ")

while guess != password:
  print("Incorrect password. Try again.")
  guess = input("Enter the password: ")

print("Access Granted.")
```

```python
num = int(input("Enter a number 1 to 10: "))

while num < 1 or num > 10:
  print("Invalid number. Try again.")
  num = int(input("Enter a number 1 to 10: "))

print(f"You entered a valid number: {num}")
```

```python
total = 0
user_num = int(input("Enter a number or 0 to stop: "))

while user_num != 0:
  total += user_num
  user_num = int(input("Enter a number or 0 to stop: "))

print(f"Total: {total}")
```

```python
secret_word = "python"
attempts = 0
max_attempts = 3

while attempts < max_attempts:
  guess = input("Enter the word: ")
  attempts += 1

  if guess == secret_word:
    print("Correct!")
    break
else:
    print("No attempts remaining.")
```
