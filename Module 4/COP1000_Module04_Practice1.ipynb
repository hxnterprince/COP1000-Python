```python
num = int(input("Enter a number: "))
for i in range(1, 11):
  result = num * i
  print(f"{num} * {i} = {result}")
```

```python
total = 0
for i in range(1, 11):
  total += i
print(f"The total is {total}.")
```

```python
total = 0
for i in range(1,6):
    score = float(input(f"Enter test score {i}: "))
    total += score
average = total / 5
print(f" Total: {total:.2f}")
print(f" Average: {average:.2f}")
```
