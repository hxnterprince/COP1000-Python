```python
def double_number(number):
    return number * 2

result = double_number(8)
print(result)

result2 = double_number(14)
print(result2)

result3 = double_number(80)
print(result3)
```

```python
def add_numbers(numbers1, numbers2):
    return number1 + number2

num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))

total_sum = add_numbers(num1, num2)

print(f"Sum: {total_sum}")
```

```python
def calculate_area(length, width):
    return length * width

user_length = int(input("Enter the length: "))
user_width = int(input("Enter the width: "))

area_result = calculate_area(user_length, user_width)

print(f"Area: {area_result}")
```

```python
def fahrenheit_to_celsius(fahrenheit):
    celsius = (fahrenheit - 32) * 5 / 9
    return celsius

user_fahrenheit = float(input("Enter the temperature in Fahrenheit: "))

celsius_result = fahrenheit_to_celsius(user_fahrenheit)

print(f"Tempterature in Celsius: {celsius_result:}")
```

```python
def get_letter_grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"

user_score = int(input("Enter a numeric grade: "))
assigned_grade = get_letter_grade(user_score)
print(f"Returned letter grade: {assigned_grade}")

print("Testing with three diferent scores")
print(f"Returned letter grade: Score 91 {get_letter_grade(91)}")
print(f"Returned letter grade: Score 76 {get_letter_grade(76)}")
print(f"Returned letter grade: Score 40 {get_letter_grade(40)}")
```

```python
def calculate_total(price, quantity):
    return price * quantity

def display_total(item, total):
    print(f"Total cost for {item}: ${total}")

item_name = input("Enter the name of the item: ")
item_price = float(input("Enter the price of the item: "))
item_quantity = int(input("Enter the quantity of the item: "))

total_cost = calculate_total(item_price, item_quantity)
display_total(item_name, total_cost)
```
