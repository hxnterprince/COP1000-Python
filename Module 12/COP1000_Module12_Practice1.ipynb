```python
student = {
    "name": "Jordan",
    "major": "Information Technology",
    "credits": 36
}

print("Name:", student["name"])
print("Major:", student["major"])
print("Credits:", student["credits"])
```

```python
weather = {
    "city": "Jacksonville",
    "current": {
        "temperature": 82,
        "humidity": 68,
        "wind_speed": 9
    }
}

print(f"City: {weather['city']}")
print(f"Temperature: {weather['current']['temperature']}")
print(f"Humidity: {weather['current']['humidity']}")
print(f"Wind Speed: {weather['current']['wind_speed']}")
```

```python
forecast = {
    "city": "Jacksonville",
    "temperatures": [82, 84, 86, 83, 79]
}

temps = forecast["temperatures"]

print("Temperatures:", temps)
print("First temperature:", temps[0])
print("Last temperature:", temps[-1])
print("Highest temperature:", max(temps))
print("Lowest temperature:", min(temps))
```

```python
data = {
    "cities": [
        {"name": "Jacksonville", "temperature": 82},
        {"name": "Orlando", "temperature": 87},
        {"name": "Tallahassee", "temperature": 84}
    ]
}

for city in data["cities"]:
    print(f"{city['name']}: {city['temperature']}")
```

```python
import json

json_text = '{"name": "Alex", "age": 22, "major": "Computer Science"}'

student_data = json.loads(json_text)

print("Name:", student_data["name"])
print("Age:", student_data["age"])
print("Major:", student_data["major"])
print("Type:", type(student_data))
```

```python
import json

course = {
    "course": "COP 1000",
    "language": "Python",
    "active": True
}

json_output = json.dumps(course)

print(json_output)
print("Type:", type(json_output))
```
