```python
import csv

new_name = input("Enter student name: ")
new_major = input("Enter student major: ")
new_credits = input("Enter student credits: ")

with open("student.csv", "a", newline="") as file:
    writer = csv.writer(file)

    writer.writerow([new_name, new_major, new_credits])

print("\nUpdated CSV File:")
with open("student.csv", "r") as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)
```

```python
import csv

total_credits = 0
record_count = 0

with open("student.csv", "r") as file:
    reader = csv.reader(file)

    next(reader)

    for row in reader:

        if len(row) < 3:
            continue
        credits = int(row[2])
        total_credits += credits
        record_count += 1

average_credits = total_credits / record_count

print(f"Student Records: {record_count}")
print(f"Total credits: {total_credits}")
print(f"Average credits: {average_credits}")
```
