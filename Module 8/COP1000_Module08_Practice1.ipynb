```python
note1 = input("Enter note 1: ")
note2 = input("Enter note 2: ")
note3 = input("Enter note 3: ")

with open("notes.txt", "w") as file:
    file.write(note1 + "\n")
    file.write(note2 + "\n")
    file.write(note3 + "\n")

print("\nSaved Notes:")
with open("notes.txt", "r") as file:
    saved_contents = file.read()
    print(saved_contents)
```
