```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

book1 = Book("The Hobbit", "J.R.R. Tolkien")
book2 = Book("Donkey's Love Story", "Shrek")

print(f"Book 1 -> Title: {book1.title}, Author: {book1.author}")
print(f"Book 2 -> Title: {book2.title}, Author: {book2.author}")
```

```python
class Book:
    def __init__(self, title, author, year, available):
        self.title = title
        self.author = author
        self.year = year
        self.available = available

book1 = Book("Python Basics", "A. Smith", 2025, True)
book2 = Book("Data Structures", "B. Jones", 2023, False)

print(f"Title: {book1.title}")
print(f"Author: {book1.author}")
print(f"Year: {book1.year}")
print(f"Available: {book1.available}")
```

```python
class Product:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

prod = Product("Laptop", 999.99, 10)

print(f"Original Quantity: {prod.quantity}")

prod.quantity = 7

print(f"Updated Quantity: {prod.quantity}")
```

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

book1 = Book("The Hobbit", "J.R.R. Tolkien")
book2 = Book("Donkey's Love Story", "Shrek")
book3 = Book("Minecraft Dungeons", "Steve")

books = [book1, book2, book3]

for book in books:
    print(f"Title: {book.title}, Author: {book.author}")
```

```python
class Movie:

    def __init__(self, title, director, release_year):
        self.title = title
        self.director = director
        self.release_year = release_year

movie1 = Movie("Inception", "Christopher Nolan", 2010)
movie2 = Movie("Spirited Away", "Hayao Miyazaki", 2001)

print(f"Movie: {movie1.title} # Director: {movie1.director} # Year: {movie1.release_year}")
print(f"Movie: {movie2.title} # Director: {movie2.director} # Year: {movie2.release_year}")
```
