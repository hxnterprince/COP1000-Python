```python
class Student:
    def __init__(self, name, major):
        self.name = name
        self.major = major

    def display_info(self):
        print(f"Student: {self.name}")
        print(f"Major: {self.major}")

student1 = Student("Don", "Donut Maker")

student1.display_info()
```

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def calculate_area(self):
        return self.length * self.width

rect1 = Rectangle(5, 10)
rect2 = Rectangle(7, 3)

print(f"Rectangle 1 Area: {rect1.calculate_area()}")
print(f"Rectangle 2 Area: {rect2.calculate_area()}")
```

```python
class BankAccount:
    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

account = BankAccount("Alex", 500.00)

print(f"Starting Balance: ${account.balance:.2f}")

deposit_amount = 100.00
account.deposit(deposit_amount)
print(f"Deposit: ${deposit_amount:.2f}")

print(f"New Balance: ${account.balance:.2f}")
```

```python
class BankAccount:
    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient funds.")
        else:
            self.balance -= amount
            print(f"Successfully withdrew ${amount:.2f}")

account = BankAccount("Alex", 500.00)

print(f"Current Balance: ${account.balance:.2f}")
print("Attempting to withdraw $600.00...")
account.withdraw(600.00)

print(f"Current Balance: ${account.balance:.2f}")
print("Attempting to withdraw $200.00...")
account.withdraw(200.00)
print(f"New Balance: ${account.balance:.2f}")
```

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.availability = True

    def check_out(self):
        if self.availability:
            self.availability = False
            print(f"'{self.title}' has been successfully checked out.")
        else:
            print(f"Sorry, '{self.title}' is already checked out.")

    def return_book(self):
        self.availability = True
        print(f"'{self.title}' has been successfully returned.")

    def display_status(self):
        status = "Available" if self.availability else "Checked Out"
        print(f"Title: {self.title} | Status: {status}")

my_book = Book("Python Basics", "A. Smith")
my_book.display_status()

my_book.check_out()
my_book.display_status()

my_book.return_book()
my_book.display_status()
```

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.availability = True

    def check_out(self):
        self.availability = False

    def display_info(self):
        status = "Available" if self.availability else "Checked Out"
        print(f"'{self.title}' by {self.author} - [{status}]")

b1 = Book("The Hobbit", "J.R.R. Tolkien")
b2 = Book("Donkey's Love Story", "Shrek")
b3 = Book("Stan", "Steve")

books = [b1, b2, b3]

print("--- Initial Book List ---")
for book in books:
    book.display_info()

print("\nChecking out 'Donkey's Love Story'...")
b2.check_out()

print("\n--- Updated Book List ---")
for book in books:
    book.display_info()
```

```python
class ShoppingCartItem:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    def display_item(self):
        print(f"Item: {self.name} | Price: ${self.price:.2f} | Qty: {self.quantity}")

    def calculate_total(self):
        return self.price * self.quantity

    def update_quantity(self, new_qty):
        self.quantity = new_qty
        print(f"Quantity updated for {self.name}!")

item = ShoppingCartItem("Ford Mustang", 2500000.00, 2)

item.display_item()
print(f"Subtotal: ${item.calculate_total():.2f}\n")

item.update_quantity(4)
item.display_item()
print(f"New Subtotal: ${item.calculate_total():.2f}")
```
